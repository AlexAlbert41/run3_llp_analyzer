```python
import numpy as np
import pandas as pd
import uproot
import matplotlib.pyplot as plt
import sys
sys.path.insert(0,"../")
import mplhep as hep
import pickle
import glob
import ROOT
import coffea
import awkward as ak
from coffea import hist, processor
from coffea.nanoevents.methods import candidate
from coffea.nanoevents.methods import vector
```

```python
import pickle as pkl
import os
plt.style.use(hep.style.CMS)
```

```python
pickle_filename = "HMT_Effciencies_CoffeaHists_2024_082727.pkl"
output_directory = "HMT_DPS_Plots_Test_082727/"
os.makedirs(output_directory, exist_ok=True)
bins_inner = hist.Bin("cluster_size", "Cluster Size", np.concatenate([np.arange(0, 500, 100), np.arange(500, 1000, 400)]))
bins_outer = hist.Bin("cluster_size", "Cluster Size", np.concatenate([np.arange(0, 200, 50), np.arange(200, 1000, 100)]))
inner_threshold = 500
outer_threshold = 200
xlim_inner = 900
xlim_outer = 800
```

```python
with open (pickle_filename, "rb") as r:
    coffea_hists_load = pkl.load(r)
```

```python
plt.figure(figsize=(14,10))
for station_ring in list(coffea_hists_load.keys()):
    #print(station_ring)
    plt.clf()
    loc='best'
    if "denom" in station_ring: continue
    if int(station_ring[3])==1:
        bins = bins_inner
        threshold = inner_threshold
        xlim = xlim_inner
    else:
        bins = bins_outer
        threshold = outer_threshold
        xlim = xlim_outer
    if int(station_ring[3])==3:
        loc = 'lower right'
    hist.plotratio(
        ax=plt.gca(),
        num =coffea_hists_load[station_ring].project(bins),
        denom=coffea_hists_load[station_ring.replace("num", "denom")].project(bins),
        error_opts={'marker': '.', 'markersize': 20, 'color':'black'},
        unc='clopper-pearson',
        label= "Data 2022+2023",
        clear=False,
        xerr=True
        )
    plt.axvline(threshold, color="purple", label="HLT Threshold", linewidth=4, ymax=0.85)
    handles, labels = plt.gca().get_legend_handles_labels()
    plt.legend(title=station_ring[:3]+'/'+station_ring[3], loc=loc)
    plt.xlabel("$\mathregular{N_{hits}}$", fontsize=30)
    plt.ylabel("L1T efficiency", fontsize=30)
    plt.tick_params(axis='both', which='major', labelsize=28)
    plt.gca().text(1.0, 1.07, "23.1 $\mathregular{fb^{-1}}$(2022) + 27.7 $\mathregular{fb^{-1}}$(2023) (13.6 TeV)", transform=plt.gca().transAxes,
            fontsize=24, verticalalignment='top', horizontalalignment='right')
    plt.gca().text(0.02, 0.95, r"$\bf{CMS}$ $\it{Supplementary}$", transform=plt.gca().transAxes, fontsize=34, verticalalignment='top', horizontalalignment='left')
    plt.xlim(0, xlim)
    plt.ylim(0, 1.25)
    plt.savefig(output_directory+station_ring.replace('_num','')+".pdf")
```
